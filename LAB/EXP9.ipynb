### Experiment 9: Rule-Based Part-of-Speech Tagging using Regular Expressions
This program implements a rule-based POS tagger using regular expressions to classify words based on morphological clues and patterns.

```python
import nltk
from nltk.tokenize import word_tokenize

# Ensure required tokenizers are downloaded
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

# 1. Define regular expression rules for POS tagging
patterns = [
    (r'.*ing$', 'VBG'),               # gerunds / present participles (e.g., walking)
    (r'.*ed$', 'VBD'),                # simple past tense verbs (e.g., walked)
    (r'.*es$', 'VBZ'),                # 3rd person singular present verbs (e.g., reaches)
    (r'.*ould$', 'MD'),               # modals (e.g., would, could, should)
    (r'.*\'s$', 'POS'),               # possessive ending (e.g., John's)
    (r'.*s$', 'NNS'),                 # plural nouns (e.g., dogs)
    (r'^-?[0-9]+(\.[0-9]+)?$', 'CD'), # cardinal numbers (e.g., 12.5, 42)
    (r'^(The|the|A|a|An|an)$', 'DT'), # determiners
    (r'.*ly$', 'RB'),                 # adverbs (e.g., quickly)
    (r'.*', 'NN')                     # nouns (default / fallback)
]

# 2. Initialize NLTK's RegexpTagger with the patterns
regexp_tagger = nltk.RegexpTagger(patterns)

# 3. Test text
test_sentence = "The quickly running dogs barked at 12 patients."
tokens = word_tokenize(test_sentence)

# 4. Perform tagging
tagged_tokens = regexp_tagger.tag(tokens)

print("=== Experiment 9: Rule-Based POS Tagging ===")
print(f"Original Sentence: \"{test_sentence}\"\n")
print(f"{'Word':12} | {'Predicted Tag':15} | {'Description'}")
print("-" * 50)

# Map tags to simplified human-readable descriptions for display
tag_desc = {
    'VBG': 'Verb, Gerund',
    'VBD': 'Verb, Past Tense',
    'VBZ': 'Verb, Present 3rd Person',
    'MD': 'Modal',
    'POS': 'Possessive',
    'NNS': 'Noun, Plural',
    'CD': 'Cardinal Number',
    'DT': 'Determiner',
    'RB': 'Adverb',
    'NN': 'Noun, Singular (Default)'
}

for word, tag in tagged_tokens:
    desc = tag_desc.get(tag, "Punctuation/Other")
    print(f"{word:12} | {tag:15} | {desc}")
```
