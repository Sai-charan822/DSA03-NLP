### Experiment 4: Finite-State Machine for Morphological Parsing (Pluralization)
This program demonstrates a simple finite-state machine (FSM) that generates the plural form of English nouns based on their morphological ending rules.

```python
def pluralize_fsm(noun):
    # Rule 1: Nouns ending in 's', 'x', 'z', 'ch', or 'sh' -> add 'es'
    if noun.endswith(('s', 'x', 'z', 'ch', 'sh')):
        return noun + 'es'

    # Rule 2: Nouns ending in 'y' preceded by a consonant -> change 'y' to 'ies'
    elif noun.endswith('y') and len(noun) > 1 and noun[-2] not in 'aeiou':
        return noun[:-1] + 'ies'

    # Rule 3: Regular nouns -> add 's'
    else:
        return noun + 's'

# Test the morphological parser
test_nouns = ["cat", "dog", "church", "box", "bus", "baby", "toy", "quiz", "wish"]

print("=== Experiment 4: FSM for Pluralization ===")
for noun in test_nouns:
    plural = pluralize_fsm(noun)
    print(f"Singular: {noun:10} -> Plural: {plural}")
```

### Experiment 5: Porter Stemmer Algorithm
This program demonstrates how to use NLTK's Porter Stemmer algorithm to find the root/stem of a list of words.

```python
from nltk.stem import PorterStemmer

# Initialize the Porter Stemmer
stemmer = PorterStemmer()

# List of words to stem
words_list = [
    "connect", "connected", "connecting", "connection", "connections",
    "argue", "argued", "arguing", "argument", "arguments",
    "happy", "happier", "happiest",
    "study", "studying", "studies", "studied"
]

print("=== Experiment 5: Porter Stemmer ===")
print(f"{'Original Word':15} -> {'Stemmed Word'}")
print("-" * 35)
for word in words_list:
    stemmed = stemmer.stem(word)
    print(f"{word:15} -> {stemmed}")
```
