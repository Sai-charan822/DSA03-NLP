### 21. Syntax-Driven Semantic Analysis
We will combine Part-of-Speech (POS) tagging and shallow chunking to extract Noun Phrases (syntax), and then lookup their dictionary definitions and meanings using WordNet (semantics).

```python
import nltk
from nltk.corpus import wordnet as wn
from nltk.tokenize import word_tokenize

# Ensure necessary resources are downloaded
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)
nltk.download('wordnet', quiet=True)

# Define the input sentence
sentence = "The bright student solved a difficult computer science problem."

print(f"Input Sentence: \"{sentence}\"\n")

# Step 1: Tokenization & POS Tagging
tokens = word_tokenize(sentence)
pos_tags = nltk.pos_tag(tokens)
print("POS Tags:", pos_tags, "\n")

# Step 2: Syntax Chunking to find Noun Phrases (NP)
# Rule: optional determiner (DT), any number of adjectives (JJ), then nouns (NN/NNS/NNP)
grammar = "NP: {<DT>?<JJ>*<NN.*>+}"
cp = nltk.RegexpParser(grammar)
parsed_tree = cp.parse(pos_tags)

# Step 3: Extract Noun Phrases and Perform Semantic Analysis using WordNet
print("Extracted Noun Phrases & Semantic Meanings:")
print("=" * 65)

for subtree in parsed_tree.subtrees():
    if subtree.label() == 'NP':
        # Join the words of the noun phrase
        np_words = [word for word, tag in subtree.leaves()]
        np_text = " ".join(np_words)

        # Identify potential nouns inside the phrase to resolve semantics
        nouns_in_phrase = [word for word, tag in subtree.leaves() if tag.startswith('NN')]

        print(f"Noun Phrase (Syntax): \"{np_text}\"")
        print(f"  -> Key Noun(s)      : {nouns_in_phrase}")

        # Semantic Lookup
        for noun in nouns_in_phrase:
            synsets = wn.synsets(noun, pos=wn.NOUN)
            if synsets:
                # Get the primary/first sense
                primary_sense = synsets[0]
                print(f"  -> Semantic Sense  : {primary_sense.name()}")
                print(f"     Definition      : {primary_sense.definition()}")
            else:
                print(f"  -> Semantic Sense  : No WordNet definition found for '{noun}'")
        print("-" * 65)
```
