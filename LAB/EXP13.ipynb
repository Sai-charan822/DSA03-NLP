### Parsing and Visualizing Parse Trees with NLTK

We will define a Context-Free Grammar (CFG) using `nltk.CFG.fromstring` and use a `RecursiveDescentParser` or a `ChartParser` to generate and print the hierarchical parse tree.

```python
import nltk
from nltk import CFG

# Define a simple context-free grammar
# NP: Noun Phrase, VP: Verb Phrase, PP: Prepositional Phrase
# Det: Determiner, N: Noun, V: Verb, P: Preposition
grammar = CFG.fromstring("""
    S -> NP VP
    PP -> P NP
    NP -> Det N | Det N PP | 'I'
    VP -> V NP | VP PP
    Det -> 'an' | 'my' | 'a' | 'the'
    N -> 'elephant' | 'pajamas' | 'shot'
    V -> 'shot'
    P -> 'in'
""")

# Create a chart parser
parser = nltk.ChartParser(grammar)

# Define a tokenized sentence to parse
sentence = "I shot an elephant in my pajamas".split()
print("Sentence:", sentence)
```

```python
# Parse the sentence and display the parse trees
trees = list(parser.parse(sentence))

print(f"Found {len(trees)} parse tree(s):\n")
for idx, tree in enumerate(trees, 1):
    print(f"--- Parse Tree {idx} ---")
    # Print the tree representation in ASCII text format
    tree.pretty_print()
    print()
```
