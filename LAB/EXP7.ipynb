### Experiment 7: Part-of-Speech (POS) Tagging with NLTK
This program demonstrates how to use the NLTK library to tokenize a sentence and tag each word with its appropriate grammatical part of speech.

```python
import nltk
from nltk.tokenize import word_tokenize

# Ensure required NLTK datasets are downloaded
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)
nltk.download('averaged_perceptron_tagger', quiet=True)
nltk.download('averaged_perceptron_tagger_eng', quiet=True)
nltk.download('tagsets', quiet=True)

# Sample text
text = "We are learning Natural Language Processing to build intelligent systems."

# 1. Tokenize the text
tokens = word_tokenize(text)

# 2. Perform POS tagging
pos_tagged_words = nltk.pos_tag(tokens)

print("=== Experiment 7: POS Tagging Results ===")
print(f"Original Text: \"{text}\"\n")
print(f"{'Word':15} | {'POS Tag':10} | {'Definition'}")
print("-" * 60)

# Display each word, its tag, and brief explanation
for word, tag in pos_tagged_words:
    # Lookup POS tag definition using nltk.help.upenn_tagset
    try:
        # Retrieve the definition from the NLTK documentation data
        definition = nltk.help.upenn_tagset(tag, silent=True) or ""
        # Clean up the output string to keep it short
        definition = definition.split('\n')[0] if '\n' in definition else definition
    except Exception:
        definition = "N/A"

    print(f"{word:15} | {tag:10} | {definition}")
```
