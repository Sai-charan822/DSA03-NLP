### Experiment 6: Bigram Language Model for Text Generation
This program demonstrates how to build a simple bigram language model from a sample text corpus and use it to generate new text sequence predictions.

```python
import random
from collections import defaultdict

class BigramGenerator:
    def __init__(self):
        # Dict mapping a word to a list of words that can follow it
        self.model = defaultdict(list)

    def fit(self, text):
        # Simple preprocessing and tokenization
        words = text.lower().split()

        # Build bigrams
        for i in range(len(words) - 1):
            current_word = words[i]
            next_word = words[i+1]
            self.model[current_word].append(next_word)

    def generate(self, seed_word, length=10):
        word = seed_word.lower()
        if word not in self.model:
            return f"Word '{seed_word}' not found in the corpus model."

        sentence = [word]

        for _ in range(length - 1):
            next_words = self.model[word]
            if not next_words:
                break  # Stop generating if we reach a dead-end word

            # Randomly select the next word based on transition frequencies
            word = random.choice(next_words)
            sentence.append(word)

        return " ".join(sentence)

# Sample text corpus
corpus = """
Natural language processing is a subfield of linguistics, computer science, and artificial intelligence
concerned with the interactions between computers and human language, in particular how to program computers
to process and analyze large amounts of natural language data. Challenges in natural language processing
frequently involve speech recognition, natural-language understanding, and natural-language generation.
"""

# Initialize and train the bigram generator
generator = BigramGenerator()
generator.fit(corpus)

# Generate text starting with different seeds
seeds = ["natural", "language", "processing", "computers"]

print("=== Experiment 6: Bigram Text Generation ===\n")
for seed in seeds:
    generated_text = generator.generate(seed, length=12)
    print(f"Seed: '{seed}' -> \nGenerated: \"{generated_text}\"\n")
```
