### Experiment 10: Transformation-Based Tagging (Brill Tagging)
This program demonstrates a simple rule-based, transformation-based tagger. It starts with an initial baseline tagger (all nouns) and applies logical contextual rules to improve and refine the Part-of-Speech tagging accuracy.

```python
# 1. Sample lexicon containing common words and their typical baseline tag
lexicon = {
    "the": "DET", "a": "DET",
    "can": "VERB", "will": "VERB",
    "dog": "NOUN", "cat": "NOUN", "run": "NOUN", "fish": "NOUN"
}

# 2. Define transformation rules:
# A rule is represented as a dictionary:
# (current_tag, context_word, context_position, new_tag)
transformation_rules = [
    # Rule 1: Change tag from NOUN to VERB if the previous word is a Modal Verb/Auxiliary like "can"
    {
        "current_tag": "NOUN",
        "new_tag": "VERB",
        "context_word": "can",
        "offset": -1
    }
]

# 3. Simple Transformation-Based Tagger class
class SimpleTransformationTagger:
    def __init__(self, lexicon, rules):
        self.lexicon = lexicon
        self.rules = rules

    def tag(self, sentence):
        words = [w.lower() for w in sentence.split()]

        # Step 1: Baseline Tagging (Initial State)
        # Lookup tag in lexicon, default to "NOUN" if unknown
        tagged_sentence = []
        for word in words:
            initial_tag = self.lexicon.get(word, "NOUN")
            tagged_sentence.append([word, initial_tag])

        print("Initial Baseline Tagging:")
        print(tagged_sentence, "\n")

        # Step 2: Apply transformation rules to correct errors
        for rule in self.rules:
            for i in range(len(words)):
                current_word, current_tag = tagged_sentence[i]

                # Check if current tag matches the trigger condition
                if current_tag == rule["current_tag"]:
                    # Check context offset (e.g., -1 for the previous word)
                    context_idx = i + rule["offset"]
                    if 0 <= context_idx < len(words):
                        context_word = tagged_sentence[context_idx][0]

                        # If the context word matches, apply the transformation
                        if context_word == rule["context_word"]:
                            tagged_sentence[i][1] = rule["new_tag"]
                            print(f"Applying Rule -> Change '{current_word}' to '{rule['new_tag']}' because previous word is '{context_word}'")

        return [tuple(item) for item in tagged_sentence]

# 4. Run and test the tagger
test_sentence = "The dog can run"
tagger = SimpleTransformationTagger(lexicon, transformation_rules)
final_tags = tagger.tag(test_sentence)

print("\n=== Experiment 10: Final Tagged Sentence ===")
print(final_tags)
```
