### 18. First-Order Predicate Calculus (FOPC) Parser
We will implement a logic parser using `nltk.sem.Expression` to parse and inspect basic first-order logical expressions containing predicates, connectives, and quantifiers.

```python
import nltk
from nltk.sem import Expression

# Instantiate a read parser for logical expressions
read_expr = Expression.fromstring

# Define some basic FOPC expressions to parse
expressions = [
    "all x. (man(x) -> mortal(x))",
    "exists x. (dog(x) & friendly(x))",
    "love(john, mary)",
    "-happy(paul)"
]

print("Parsing First-Order Predicate Calculus (FOPC) Expressions:\n")
for expr_str in expressions:
    try:
        # Parse the string representation into an NLTK Logic Expression object
        parsed_expr = read_expr(expr_str)
        print(f"Original String : {expr_str}")
        print(f"Parsed Object   : {parsed_expr}")
        print(f"Expression Type : {type(parsed_expr).__name__}")
        print(f"Free Variables  : {parsed_expr.free()}")
        print(f"Constants/Preds : {parsed_expr.constants()}")
        print("-" * 50)
    except Exception as e:
        print(f"Error parsing '{expr_str}': {e}")
```
