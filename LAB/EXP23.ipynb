```python
import nltk
from sentence_transformers import SentenceTransformer, util

nltk.download('punkt')
nltk.download('punkt_tab')

def evaluate_coherence(text):
    # Split text into sentences
    sentences = nltk.sent_tokenize(text)
    if len(sentences) <= 1:
        return 1.0, ["Text is too short to evaluate coherence between sentences."]

    # Load a lightweight sentence transformer model
    model = SentenceTransformer('all-MiniLM-L6-v2')

    # Encode sentences to get their embeddings
    embeddings = model.encode(sentences, convert_to_tensor=True)

    similarities = []
    details = []

    # Calculate cosine similarity between consecutive sentences
    for i in range(len(sentences) - 1):
        sim = util.cos_sim(embeddings[i], embeddings[i+1]).item()
        similarities.append(sim)
        details.append(f"Similarity between sentence {i+1} and {i+2}: {sim:.4f}")

    # Average similarity represents the overall coherence score
    avg_coherence = sum(similarities) / len(similarities)
    return avg_coherence, details

# Sample texts to test
coherent_text = """Python is a popular programming language. It is widely used for data science and machine learning. Many developers prefer it because of its simple syntax and readability."""

incoherent_text = """Python is a popular programming language. The weather today is very sunny and warm. I should probably buy some apples from the grocery store."""

print("--- Evaluating Coherent Text ---")
score, steps = evaluate_coherence(coherent_text)
print(f"Overall Coherence Score: {score:.4f}")
for step in steps:
    print(step)

print("\n--- Evaluating Incoherent Text ---")
score_inc, steps_inc = evaluate_coherence(incoherent_text)
print(f"Overall Coherence Score: {score_inc:.4f}")
for step in steps_inc:
    print(step)
```
