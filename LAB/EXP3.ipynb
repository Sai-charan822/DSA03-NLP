### Experiment 3: Morphological Analysis using NLTK
This program demonstrates basic morphological analysis (Tokenization, POS tagging, Stemming, and Lemmatization) using the Natural Language Toolkit (NLTK) in Python.

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download necessary NLTK datasets
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

# Sample text
sample_text = "The cats are running quickly and studies show they enjoyed studying yesterday."
print(f"Original Text: {sample_text}\n")

# 1. Tokenization
tokens = word_tokenize(sample_text)
print(f"Tokens: {tokens}\n")

# 2. Stemming (Porter Stemmer)
stemmer = PorterStemmer()
stems = [stemmer.stem(token) for token in tokens]
print(f"Stemmed Words: {stems}\n")

# 3. Lemmatization (WordNet Lemmatizer)
lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(token) for token in tokens]
print(f"Lemmatized Words (default/Noun): {lemmas}\n")

# Lemmatization with POS context (Verbs)
lemmas_verbs = [lemmatizer.lemmatize(token, pos='v') for token in tokens]
print(f"Lemmatized Words (Verbs context): {lemmas_verbs}\n")

# 4. Part-of-Speech (POS) Tagging
pos_tags = nltk.pos_tag(tokens)
print("POS Tagging Results:")
for word, tag in pos_tags:
    print(f"  {word:12} -> {tag}")
```
