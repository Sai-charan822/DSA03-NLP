### 16. Named Entity Recognition (NER) using SpaCy
We will download and load the small English NLP model from SpaCy, process a sample text, extract named entities, and visualize them.

```python
!python -m spacy download en_core_web_sm -q

import spacy
from spacy import displacy

# Load the English NLP model
nlp = spacy.load("en_core_web_sm")

# Sample text for NER
text = (
    "Google was founded on September 4, 1998, by Larry Page and Sergey Brin while they "
    "were Ph.D. students at Stanford University in California. Today, Google is headquartered "
    "in Mountain View and is one of the most valuable companies in the world, competing "
    "with Apple and Microsoft."
)

# Process the text
doc = nlp(text)

# Print found entities
print("Extracted Entities:")
print(f"{'Entity':<25} | {'Label':<15} | {'Description'}")
print("-" * 70)
for ent in doc.ents:
    print(f"{ent.text:<25} | {ent.label_:<15} | {spacy.explain(ent.label_)}")

# Render beautiful visualization
print("\nEntity Visualization:")
displacy.render(doc, style="ent", jupyter=True)
```
