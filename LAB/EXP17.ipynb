### 17. Accessing WordNet with NLTK
We will download the WordNet dataset from NLTK, retrieve synsets for a specific word, and explore its definitions, examples, and synonyms (lemmas).

```python
import nltk
from nltk.corpus import wordnet as wn

# Download the WordNet resource
nltk.download('wordnet', quiet=True)

# Define a word to look up
word = "travel"

# Retrieve all synsets (synonym sets) for the word
synsets = wn.synsets(word)

print(f"WordNet Search for: '{word}'")
print(f"Found {len(synsets)} synsets.\n")

for i, synset in enumerate(synsets, 1):
    print(f"{i}. Synset: {synset.name()}")
    print(f"   - Part of Speech: {synset.pos()}")
    print(f"   - Definition: {synset.definition()}")

    # Display example sentences if they exist
    examples = synset.examples()
    if examples:
        print(f"   - Examples: {examples}")

    # Retrieve lemmas (synonymous words belonging to this synset)
    lemmas = [lemma.name() for lemma in synset.lemmas()]
    print(f"   - Synonyms/Lemmas: {lemmas}")
    print("-" * 60)
```
